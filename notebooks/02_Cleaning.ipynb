```python
# ============================================
# 1. Introduction
# ============================================

# This notebook cleans the extended Telco Customer Churn dataset.
# It includes:
# - Data type correction
# - Handling of missing values
# - Category standardization
# - Removal of irrelevant columns
# - Export of the clean dataset

# ============================================
# 2. Load libraries
# ============================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")
sns.set(style="whitegrid")

# ============================================
# 3. Load raw dataset
# ============================================

df = pd.read_csv("../data/raw/Telco-Customer-Churn.csv")
df.head()

# Initial dimensions
df.shape

# ============================================
# 4. Missing values review
# ============================================

df.isnull().sum().sort_values(ascending=False)

# Missing values heatmap
plt.figure(figsize=(12,6))
sns.heatmap(df.isnull(), cbar=False)
plt.title("Missing Values Map")
plt.show()

# Convert numeric columns that were loaded incorrectly
numeric_cols = [
    "MonthlyCharge", "TotalCharges", "TotalRefunds",
    "TotalExtraDataCharges", "TotalLongDistanceCharges",
    "TotalRevenue", "CLTV", "ChurnScore", "SatisfactionScore",
    "TenureinMonths", "AvgMonthlyGBDownload",
    "AvgMonthlyLongDistanceCharges"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Simple median imputation
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

categorical_cols = df.select_dtypes(include=["object"]).columns

for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip()

cols_to_drop = [
    "CustomerID", "Country", "State", "City",
    "ZipCode", "Latitude", "Longitude", "Population", "Quarter"
]

df.drop(columns=cols_to_drop, inplace=True)

df.drop_duplicates(inplace=True)
df.shape

df.info()

df.to_csv("../data/processed/clean_telco.csv", index=False)
print("Clean dataset saved to data/processed/clean_telco.csv")

print("""
CLEANING CONCLUSIONS:

1. Numeric columns that were stored as text were corrected.
2. Missing values in numeric variables were imputed with the median.
3. Categories were standardized by removing spaces and inconsistent formats.
4. Columns irrelevant to analysis and modeling were removed.
5. Duplicates were removed.
6. The dataset is ready for Feature Engineering in 03_Feature_Engineering.ipynb.
""")
```
