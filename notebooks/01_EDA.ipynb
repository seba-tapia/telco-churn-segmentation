```python
# ============================================
# 1. Introduction
# ============================================

# This notebook performs Exploratory Data Analysis (EDA)
# on the extended Telco Customer Churn dataset.
# It covers demographic variables, services,
# billing, satisfaction, CLTV, and churn.

# ============================================
# 2. Load libraries
# ============================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")

# ============================================
# 3. Load dataset
# ============================================

df = pd.read_csv("../data/raw/Telco-Customer-Churn.csv")

df.head()

# Dataset dimensions
df.shape

# Dataset columns
df.columns

# ============================================
# 4. General dataset information
# ============================================

df.info()

# Missing values count
df.isnull().sum().sort_values(ascending=False)

# ============================================
# 5. Descriptive statistics
# ============================================

df.describe()

# ============================================
# 6. Target variable analysis
# ============================================

# ChurnLabel distribution
df["ChurnLabel"].value_counts()

# Churn percentage
df["ChurnLabel"].value_counts(normalize=True) * 100

# Churn plot
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="ChurnLabel", palette="viridis")
plt.title("Churn Distribution")
plt.show()

# ============================================
# 7. CustomerStatus
# ============================================

df["CustomerStatus"].value_counts()

plt.figure(figsize=(8,4))
sns.countplot(data=df, x="CustomerStatus", palette="magma")
plt.title("CustomerStatus Distribution")
plt.xticks(rotation=45)
plt.show()

# ============================================
# 8. Demographic variables
# ============================================

demographic_cols = ["Gender", "Age", "Under30", "SeniorCitizen", 
                    "Married", "Dependents", "NumberofDependents"]

df[demographic_cols].describe(include="all")

plt.figure(figsize=(6,4))
sns.histplot(df["Age"], kde=True, bins=30)
plt.title("Age Distribution")
plt.show()

# ============================================
# 9. Contracted services
# ============================================

service_cols = [
    "PhoneService", "InternetService", "InternetType",
    "OnlineSecurity", "OnlineBackup", "DeviceProtectionPlan",
    "PremiumTechSupport", "StreamingTV", "StreamingMovies",
    "StreamingMusic", "UnlimitedData"
]

df[service_cols].describe(include="all")

plt.figure(figsize=(10,6))
sns.heatmap(df[service_cols].apply(lambda x: x.astype("category").cat.codes).corr(),
            annot=False, cmap="coolwarm")
plt.title("Correlation Between Services")
plt.show()

# ============================================
# 10. Billing
# ============================================

billing_cols = [
    "MonthlyCharge", "TotalCharges", "TotalRevenue",
    "TotalRefunds", "TotalExtraDataCharges", "TotalLongDistanceCharges"
]

df[billing_cols].describe()

plt.figure(figsize=(6,4))
sns.histplot(df["MonthlyCharge"], kde=True)
plt.title("MonthlyCharge Distribution")
plt.show()

# Relationship between MonthlyCharge and Churn
plt.figure(figsize=(6,4))
sns.boxplot(data=df, x="ChurnLabel", y="MonthlyCharge")
plt.title("MonthlyCharge vs Churn")
plt.show()

# ============================================
# 11. Tenure
# ============================================

plt.figure(figsize=(6,4))
sns.histplot(df["TenureinMonths"], kde=True, bins=30)
plt.title("Tenure Distribution")
plt.show()

# ============================================
# 12. Satisfaction and CLTV
# ============================================

plt.figure(figsize=(6,4))
sns.histplot(df["SatisfactionScore"], kde=True)
plt.title("SatisfactionScore Distribution")
plt.show()

plt.figure(figsize=(6,4))
sns.histplot(df["CLTV"], kde=True)
plt.title("CLTV Distribution")
plt.show()

# ============================================
# 13. Correlations
# ============================================

numeric_df = df.select_dtypes(include=["int64", "float64"])

plt.figure(figsize=(14,10))
sns.heatmap(numeric_df.corr(), cmap="coolwarm", annot=False)
plt.title("Correlation Matrix (numeric variables only)")
plt.show()

# ============================================
# 14. EDA Conclusions
# ============================================

print("""
EDA CONCLUSIONS:

1. The dataset contains several enriched variables (CLTV, ChurnScore, ChurnReason).
2. The churn distribution is reasonably balanced for analysis.
3. MonthlyCharge and Tenure show a clear relationship with churn.
4. Contracted services show patterns that are useful for segmentation.
5. Some variables require cleaning and encoding.
6. The next step is to clean the dataset in 02_Cleaning.ipynb.
""")




sns.set(style="whitegrid")
```
